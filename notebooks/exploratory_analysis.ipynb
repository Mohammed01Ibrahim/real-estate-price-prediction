# Exploring the supported residential sale scope
This output-free notebook is for aggregate exploration, not model training. It uses reusable modules; the original notebooks remain in the historical ZIPs.
Open from this repository with the installed `.venv` interpreter. Notebook execution is optional; install Jupyter separately if you want an interactive notebook. Never publish listing-level output, descriptions, or exact coordinates.

```python
from real_estate.data import read_raw, prepare_data
from real_estate.train import make_splits
import matplotlib.pyplot as plt
X, y, groups, report = prepare_data(read_raw())
train_idx, val_idx, test_idx = make_splits(X, y, groups)
training_features, training_prices = X.iloc[train_idx], y.iloc[train_idx]
{key: report[key] for key in ("raw_rows", "retained_rows", "missing_features", "unused_source_fields")}
```

## Training partition only
EDA here avoids test-target inspection. Scope restrictions are fixed business/schema rules, not percentile filters. Missing room/covered-area fields cannot support the historical area ratios.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), layout="constrained")
axes[0].hist(training_prices, bins=60)
axes[0].set(xlabel="Asking price (USD; all training targets)", ylabel="Listings")
training_features.isna().mean().plot.bar(ax=axes[1])
axes[1].set(ylabel="Missing fraction", title="Training feature availability")
plt.show()
```
